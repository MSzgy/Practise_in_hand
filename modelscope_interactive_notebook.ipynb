## 使用建议

- 第一次点击运行会下载并加载模型，耗时取决于模型大小和平台缓存。
- 如果只是验证界面和链路，用默认 0.5B 模型。
- 如果要观察更接近真实部署的显存和延迟，把 `MODEL_ID` 改成 `Qwen/Qwen2.5-7B-Instruct` 后重启内核再运行。
- 如果显存不足，降低 `max_new_tokens`，或者换回 0.5B 模型。
